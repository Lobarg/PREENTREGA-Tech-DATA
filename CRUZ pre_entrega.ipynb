```python
# === NORMALIZACIÓN DE CAMPOS DE TEXTO ===

dataframes = [clientes_etl, ventas_etl, mktg_etl]

for df in dataframes:
    for col in df.select_dtypes(include=['object']).columns:
        # Normaliza texto: primera letra mayúscula, resto minúsculas
        df[col] = df[col].astype(str).str.strip().str.title()

print("Normalización completada: todas las columnas de texto tienen formato 'Primera Letra En Mayúscula'.")
```

## Ventas Mensuales (corresponde a la Etapa 1, item 2 del enunciado)
Ahora que tenemos el dato de fecha de ventas en el formato adecuado podemos Agregar la información para mostrarla

```python
# ============================================================
# CÁLCULO DE VENTAS TOTALES MENSUALES

# Calcular el importe total por registro
ventas_etl['importe_total'] = ventas_etl['precio'] * ventas_etl['cantidad']

#  Crear columna Año-Mes
ventas_etl['anio_mes'] = ventas_etl['fecha_venta'].dt.to_period('M')

# Agrupar por mes y sumar importes
ventas_mensuales = (
    ventas_etl.groupby('anio_mes')['importe_total']
    .sum()
    .reset_index()
    .sort_values('anio_mes')
)

#  Mostrar resultados
print("=== VENTAS TOTALES MENSUALES ===")
print(ventas_mensuales)

## Guardar archivo en la subcarpeta
ventas_mensuales.to_csv(f"{ruta_reportes}/ventas_totales_mensuales.csv", index=False, encoding="utf-8")

```

# Transformación de Datos (Punto 2, etapa 2)

Aplicar filtros y transformaciones para crear una tabla de ventas que muestre solo los productos con alto rendimiento.


```python
# Agrupar por producto y calcular métricas relevantes
ventas_top = (
    ventas_etl.groupby('producto')
    .agg(
        importe_total=('importe_total', 'sum'),      # total vendido (precio x cantidad)
        unidades=('cantidad', 'sum'),                # total de unidades vendidas
        precio_promedio=('precio', 'mean'),          # precio promedio unitario
        registros=('producto', 'count')              # cantidad de registros
    )
    .reset_index()
    .sort_values('importe_total', ascending=False)   # orden descendente por importe
)
# Agrupar por producto y calcular métricas relevantes
ventas_top = (
    ventas_etl.groupby('producto')
    .agg(
        importe_total=('importe_total', 'sum'),      # total vendido (precio x cantidad)
        unidades=('cantidad', 'sum'),                # total de unidades vendidas
        precio_promedio=('precio', 'mean'),          # precio promedio unitario
        registros=('producto', 'count')              # cantidad de registros
    )
    .reset_index()
    .sort_values('importe_total', ascending=False)   # orden descendente por importe
)

# Mostrar productos ordenados de mayor a menor importe total
print("\n----- PRODUCTOS ORDENADOS DE MAYOR A MENOR IMPORTE DE VENTAS ------")
print(ventas_top[['producto','importe_total']])

# Calcular el valor del percentil 80 del importe total
umbral_p80 = ventas_top['importe_total'].quantile(0.8)

# Filtrar productos que superan el percentil 80
ventas_top80 = ventas_top[ventas_top['importe_total'] > umbral_p80]

# Mostrar productos del top 20 por ciento
print("\n----- TOP PERCENTIL 80 - POR IMPORTE TOTAL DE VENTAS (Precio x Cantidad) ------")
print(ventas_top80)

# ============================================================
# GUARDAR RESULTADO EN LA CARPETA DE REPORTES
# ============================================================

ventas_top80.to_csv(
    f"{ruta_reportes}/ventas_top80_completo.csv",
    index=False,
    encoding="utf-8"
)

print(f"\nArchivo generado correctamente en: {ruta_reportes}/ventas_top80_completo.csv")

# Calcular el valor del percentil 80 del importe total
umbral_p80 = ventas_top['importe_total'].quantile(0.8)

# Filtrar productos que superan el percentil 80
ventas_top80 = ventas_top[ventas_top['importe_total'] > umbral_p80]

# Mostrar productos del top 20 por ciento
print("\n----- TOP PERCENTIL 80 - POR IMPORTE TOTAL DE VENTAS (Precio x Cantidad) ------")
print(ventas_top80)

# ============================================================
# GUARDAR RESULTADO EN LA CARPETA DE REPORTES

ventas_top80.to_csv(
    f"{ruta_reportes}/ventas_top80_completo.csv",
    index=False,
    encoding="utf-8"
)

print(f"\nArchivo generado correctamente en: {ruta_reportes}/ventas_top80_completo.csv")
```

#** Integración de datos** (Etapa 2 Item 4 del enunciado):*
Combinar los sets de datos de ventas y marketing para obtener una visión más amplia de las tendencias.

Listado completo, por registro

```python
import pandas as pd
import numpy as np

# === 1. Determinar el periodo total de análisis ===
inicio_periodo = ventas_etl['fecha_venta'].min()
fin_periodo = ventas_etl['fecha_venta'].max()
dias_periodo_total = (fin_periodo - inicio_periodo).days + 1

print("Periodo analizado:", inicio_periodo.date(), "a", fin_periodo.date(), f"({dias_periodo_total} días)")

# === 2. Convertir fechas si aún no están en formato datetime ===
mktg_etl['fecha_inicio'] = pd.to_datetime(mktg_etl['fecha_inicio'], dayfirst=True)
mktg_etl['fecha_fin'] = pd.to_datetime(mktg_etl['fecha_fin'], dayfirst=True)
ventas_etl['fecha_venta'] = pd.to_datetime(ventas_etl['fecha_venta'], dayfirst=True)

# === 3. Marcar ventas dentro de campaña SIN duplicar registros ===
ventas_etl['en_campania'] = False

for _, camp in mktg_etl.iterrows():
    mask = (
        (ventas_etl['producto'] == camp['producto']) &
        (ventas_etl['fecha_venta'] >= camp['fecha_inicio']) &
        (ventas_etl['fecha_venta'] <= camp['fecha_fin'])
    )
    ventas_etl.loc[mask, 'en_campania'] = True

# === 4. Separar ventas con y sin campaña ===
ventas_con_campania = ventas_etl[ventas_etl['en_campania'] == True].copy()
ventas_sin_campania = ventas_etl[ventas_etl['en_campania'] == False].copy()

print("Ventas con campaña:", len(ventas_con_campania))
print("Ventas sin campaña:", len(ventas_sin_campania))

# === 5. Calcular días de campaña por producto y canal ===
dias_campania = (
    mktg_etl
    .groupby(['producto', 'canal'])[['fecha_inicio', 'fecha_fin']]
    .apply(lambda x: (x['fecha_fin'].max() - x['fecha_inicio'].min()).days + 1)
    .reset_index(name='dias_campania')
)

# === 6. Agrupar ventas con campaña ===
con_campania = (
    ventas_con_campania
    .groupby(['producto'])
    .agg(cantidad=('cantidad', 'sum'), importe_total=('importe_total', 'sum'))
    .reset_index()
)

# Añadir canal y días de campaña desde marketing
con_campania = pd.merge(con_campania, dias_campania, on='producto', how='left')
con_campania['indice_venta_dia'] = con_campania['importe_total'] / con_campania['dias_campania']

# === 7. Calcular días fuera de campaña ===
dias_totales_prod = dias_campania.groupby('producto')['dias_campania'].sum().reset_index()
dias_totales_prod['dias_fuera_campania'] = dias_periodo_total - dias_totales_prod['dias_campania']
dias_totales_prod['dias_fuera_campania'] = dias_totales_prod['dias_fuera_campania'].clip(lower=1)

# === 8. Agrupar ventas sin campaña ===
sin_campania = (
    ventas_sin_campania
    .groupby('producto')
    .agg(cantidad=('cantidad', 'sum'), importe_total=('importe_total', 'sum'))
    .reset_index()
)
sin_campania = pd.merge(sin_campania, dias_totales_prod[['producto', 'dias_fuera_campania']],
                        on='producto', how='left')
sin_campania = sin_campania.rename(columns={'dias_fuera_campania': 'dias_campania'})
sin_campania['canal'] = 'Sin campaña'
sin_campania['indice_venta_dia'] = sin_campania['importe_total'] / sin_campania['dias_campania']

# === 9. Combinar resultados ===
comparativo = pd.concat([con_campania, sin_campania], ignore_index=True)

# === 10. Listado completo con días evaluados ===
print("\n=== LISTADO COMPLETO POR PRODUCTO, CANAL Y DÍAS EVALUADOS ===")
print(
    comparativo[['producto', 'canal', 'dias_campania', 'cantidad', 'importe_total', 'indice_venta_dia']]
    .sort_values(['producto', 'canal'])
    .round(2)
)
print(f"\nTotal de días analizados para todo el periodo: {dias_periodo_total}")

```

Tabla comparativa de canales

```python

# === 10. Obtener canales dinámicamente ===
canales = sorted(mktg_etl['canal'].dropna().unique().tolist() + ['Sin campaña'])

# === 11. Crear tabla comparativa ===
pivot_indice = comparativo.pivot_table(
    index='producto',
    columns='canal',
    values='indice_venta_dia',
    aggfunc='sum'
).reset_index()

# Asegurar columnas para todos los canales existentes
for canal in canales:
    if canal not in pivot_indice.columns:
        pivot_indice[canal] = 0

pivot_indice = pivot_indice[['producto'] + canales]

print("\n=== INDICE DE VENTA DIARIA ($/día) POR PRODUCTO Y CANAL ===")
print(pivot_indice.round(2))

# === 12. Promedio general por canal ===
promedios = comparativo.groupby('canal')['indice_venta_dia'].mean().round(2)
print("\n=== PROMEDIO GENERAL DEL INDICE DIARIO ($/día) POR CANAL ===")
print(promedios)

# === 13. Conclusión básica ===
canal_top = promedios.idxmax()
valor_top = promedios.max()
print(f"\nEl canal con mayor índice promedio diario de ventas es '{canal_top}' con ${valor_top:.2f} por día.")
```

**ANALISIS** El análisis conjunto de los registros de ventas y las campañas de marketing del período enero a diciembre de 2024 (364 días) evidencia una correlación positiva entre la ejecución de campañas y el aumento de las ventas diarias promedio.

Las campañas distribuidas en los distintos canales (Email, RRSS y TV) lograron incrementar significativamente el rendimiento diario respecto de las ventas sin apoyo de marketing:
